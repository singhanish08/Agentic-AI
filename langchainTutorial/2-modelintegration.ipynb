Google Gemini Model Integration

```python
import os 
from langchain.chat_models import init_chat_model
os.environ["GOOGLE_API_KEY"] = os.getenv("GOOGLE_API_KEY")
model = init_chat_model("google_genai:gemini-3.5-flash-lite")
response = model.invoke("Hello how are you?")
response.text
```

```python
import os
from langchain_google_genai import ChatGoogleGenerativeAI
model = ChatGoogleGenerativeAI(model="gemini-3.5-flash-lite")
response = model.invoke("Hello how are you?")
response.text
```

GROQ MODEL INTEGRATION

```python
import os 
from langchain.chat_models import init_chat_model
os.environ["GROQ_API_KEY"] = os.getenv("GROQ_API_KEY")
model = init_chat_model("openai/gpt-oss-20b", model_provider="groq")
response = model.invoke("Hello how are you?")
response.text
```

```python
import os 
from langchain_groq import ChatGroq
model = ChatGroq(model="openai/gpt-oss-20b")
response = model.invoke("Hello how are you?")
response.text
```

Streaming - as content is generated, it is sent to the client in chunks. This allows for faster response times and a more interactive experience.

```python
model.invoke("Write me a 200 words paragraph on Artificial Intelligence")
```

```python
for chunk in model.stream("Why do parrots have colorful feathers?"):
    print(chunk.text, end="", flush=True)
```

Batch

Batching is a technique used to process multiple requests together in a single operation. This can improve efficiency and reduce latency, especially when dealing with large volumes of data or requests.

```python
responses = model.batch(["Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"
])
for response in responses:
    print(response.text)
```

```python
model.batch(
    ["Why do parrots have colorful feathers?",
    "How do airplanes fly?",
    "What is quantum computing?"],
    config={
        'max_concurrency': 5,  # Limit to 5 parallel calls
    }
)
```
